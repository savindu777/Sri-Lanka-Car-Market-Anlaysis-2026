```python
import pandas as pd
import numpy as np
import re
```

```python
df = pd.read_csv('combined_datasets.csv')
df = df.iloc[:, :-3]
```

```python
df.head(10)
```

```python
# 2. Create a temporary cleaned series (without altering the original df['Model'] column)
cleaned_models = df['Model'].astype(str).str.strip().str.lower()

# 3. Get unique sorted model names from the temporary series
unique_models = sorted(cleaned_models.unique())

# 4. Display results
print(f"Total Unique Models (cleaned view): {len(unique_models)}\n")
print("List of Unique Models:")
for model in unique_models:
    print(f"- {model}")
```

```python
print(unique_models)
```

```python
output_txt = 'unique_models.txt'
with open(output_txt, 'w', encoding='utf-8') as f:
    for model in unique_models:
        f.write(f"{model}\n")

print(f"Successfully saved {len(unique_models)} unique models to '{output_txt}'!")
```

```python
df['Model_cleaned'] = (
    df['Model']
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r'\s+', '', regex=True)
)
```

```python
models = sorted(df['Model_cleaned'].unique())
```

```python
df['Mileage'] = (
    df['Mileage']
    .astype(str)
    .str.replace('km', '', case=False)
    .str.replace(',', '', regex=True)
    .str.strip()
)

df['Mileage_numeric'] = pd.to_numeric(df['Mileage'], errors='coerce')

# 3. Optional: Inspect the cleaned column and its data type
print(df[['Mileage_numeric']].info())
print(df['Mileage_numeric'].head(10))
```

```python
# 2. Clean and convert 'Price' to numeric:
# - Convert to string, remove 'Rs.' (case-insensitive), remove commas, and strip whitespace
# - Convert to numeric using pd.to_numeric, which automatically coerces text like 'Negotiable' into NaN
df['Price'] = (
    df['Price']
    .astype(str)
    .str.replace('Rs.', '', case=False)
    .str.replace(',', '', regex=True)
    .str.strip()
)

df['Price_numeric'] = pd.to_numeric(df['Price'], errors='coerce')

# 3. Optional: Check the result and data type
print(df[['Price_numeric']].info())
print(df['Price_numeric'].head(15))

# Save the updated dataframe if needed
# df.to_csv('cleaned_datasets.csv', index=False)
```

```python
# 2. Extract ONLY the Date portion (Year, Month, Day) and ignore time & trailing text
def extract_date_only(val):
    if pd.isna(val):
        return pd.NaT
    # Matches patterns like '2026 Jun 05' from strings like '2026 Jun 05, 11:39 am'
    match = re.search(r'(\d{4}\s+[A-Za-z]+\s+\d{2})', str(val))
    if match:
        return match.group(1)
    return str(val)

df['Ad_Date_Clean_Str'] = df['Ad Date'].apply(extract_date_only)

# Convert to standard datetime with time entirely set to 00:00:00 (pure date)
df['Ad_Date_clean'] = pd.to_datetime(df['Ad_Date_Clean_Str'], format='mixed', errors='coerce').dt.normalize()

print("Sample Date-Only Column:")
print(df[['Ad Date', 'Ad_Date_clean']].head())
```

```python
df.to_csv('formatted_data.csv', index=False)
```
